#  Display Value Counts for Categorical Columns

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
  print(f"Value counts for column: {col}")
  print(df[col].value_counts())
  print("-" * 30)
```

#  Encode Categorical Columns (Label Encoding)

```python
from sklearn.preprocessing import LabelEncoder

# Apply Label Encoding to each categorical column
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])

display(df.head())
```

#  Normalize Numeric Columns ('duration' & 'balance')

```python
from sklearn.preprocessing import MinMaxScaler

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Reshape the 'duration' column and apply MinMax scaling
df['duration'] = scaler.fit_transform(df[['duration']])
df['balance'] = scaler.fit_transform(df[['balance']])

display(df.head())
```

#  Split Dataset into Training and Testing Sets

```python
from sklearn.model_selection import train_test_split

# Define features (X) and target (y)
X = df.drop('deposit', axis=1)
y = df['deposit']

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

print("Training set shape (X_train):", X_train.shape)
print("Testing set shape (X_test):", X_test.shape)
print("Training set shape (y_train):", y_train.shape)
print("Testing set shape (y_test):", y_test.shape)
```

#  Import All Machine Learning Models

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
import pandas as pd
```

# Initialize All Models

```python
# Instantiate the models
lr_model = LogisticRegression(random_state=42)
dt_model = DecisionTreeClassifier(random_state=42)
rf_model = RandomForestClassifier(random_state=42)
knn_model = KNeighborsClassifier()
svm_model = SVC(probability=True, random_state=42)
gnb_model = GaussianNB()
mnb_model = MultinomialNB()
bnb_model = BernoulliNB()
xgb_model = XGBClassifier(random_state=42)
lgbm_model = LGBMClassifier(random_state=42)
catb_model = CatBoostClassifier(random_state=42, verbose=0)
ada_model = AdaBoostClassifier(random_state=42)

# Store models in a list
models = [
    ('Logistic Regression', lr_model),
    ('Decision Tree', dt_model),
    ('Random Forest', rf_model),
    ('KNN', knn_model),
    ('SVM', svm_model),
    ('Gaussian Naive Bayes', gnb_model),
    ('Multinomial Naive Bayes', mnb_model),
    ('Bernoulli Naive Bayes', bnb_model),
    ('XGBoost', xgb_model),
    ('LightGBM', lgbm_model),
    ('CatBoost', catb_model),
    ('AdaBoost', ada_model)
]

print("Models instantiated and stored in the 'models' list.")
```

#  Train & Evaluate All Models

```python
# Create a dictionary to store results
results = {}

# Filter out Multinomial Naive Bayes model
filtered_models = [(name, model) for name, model in models if name != 'Multinomial Naive Bayes']

# Iterate through models
for name, model in filtered_models:
    # Train the model
    model.fit(X_train, y_train)

    # Make predictions
    y_pred = model.predict(X_test)

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    roc_auc = roc_auc_score(y_test, y_pred)

    # Store results
    results[name] = {
        'Accuracy': accuracy,
        'Precision': precision,
        'Recall': recall,
        'F1 Score': f1,
        'ROC-AUC': roc_auc
    }

# Print the results
for name, metrics in results.items():
    print(f"--- {name} ---")
    for metric_name, value in metrics.items():
        print(f"{metric_name}: {value:.4f}")
    print("-" * 20)
```

#  Display Model Performance Results

```python
results_df = pd.DataFrame(results).T
display(results_df)
```
